```python
import os
import datetime

def find_files_created_within_last_minute(folder_path):
    current_time = datetime.datetime.now()
    one_minute_ago = current_time - datetime.timedelta(minutes=60)
    recent_files = []

    for file in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file)
        if os.path.isfile(file_path):
            
            creation_time = datetime.datetime.fromtimestamp(os.path.getctime(file_path))
            if creation_time > one_minute_ago:
                recent_files.append(file_path)

    return recent_files

```

```python
import os
import datetime
from glob import glob

def find_files_created_within_last_minute(folder_path):
    current_time = datetime.datetime.now()
    one_minute_ago = current_time - datetime.timedelta(minutes=1)
    recent_files = []

    for file_path in glob(folder_path):
        # print('file_path : ',file_path)
        creation_time = datetime.datetime.fromtimestamp(os.path.getctime(file_path))
        if creation_time > one_minute_ago:
            recent_files.append(file_path)

    return recent_files
```

```python


def list_json_files_and_subdirectories(root_dir):
    json_files = []
    subdirectories = []

    for root, dirs, files in os.walk(root_dir):
        for file in files:
            if file.endswith(".txt"):
                json_files.append(os.path.join(root, file))
        for dir in dirs:
            subdirectories.append(os.path.join(root, dir))

    return json_files, subdirectories

```

```python
from datetime import datetime, timedelta

def get_time_range(base_time):
    # Convert the base_time to a datetime object
    base_time_dt = datetime.strptime(base_time, '%H%M%S')

    # Calculate the previous two seconds
    prev_two_sec = [base_time_dt - timedelta(seconds=i) for i in range(1, 3)]

    # Calculate the next two seconds
    next_two_sec = [base_time_dt + timedelta(seconds=i) for i in range(1, 3)]
    final_list=[]
    final_list.extend(prev_two_sec)
    final_list.extend([base_time_dt])
    final_list.extend(next_two_sec)
    for i in range(len(final_list)):
        final_list[i]=final_list[i].strftime('%H%M%S')
    return final_list

# Example usage
base_time = '172852'#'17:28:52'
final_list = get_time_range(base_time)

# Print the results
# print("Previous two seconds:")
for time in final_list:
    print(time)
    # print(time.strftime('%H:%M:%S'))

# print("\nNext two seconds:")
# for time in next_two_sec:
#     print(time.strftime('%H:%M:%S'))
```

```python
import subprocess

def get_running_python_processes():
    # Run the shell command ps -ef | grep python3
    process = subprocess.Popen(["ps", "-ef"], stdout=subprocess.PIPE)
    output, _ = process.communicate()

    # Decode the byte string output to utf-8
    output = output.decode("utf-8")

    # Split the output into lines and filter lines containing 'python3'
    # print(output)
    python_processes = [line.split()[8] for line in output.splitlines() if 'python3' in line]

    return python_processes

running_python_processes = get_running_python_processes()
running_python_processes
```

# Most Recent File

```python
import os
import glob

def get_most_recent_file(folder_path):
    # Get list of all files in the folder
    print('input : ',folder_path)
    files = glob.glob(folder_path)#glob.glob(os.path.join(folder_path, '*'))

    if not files:
        return None

    # Find the file with the latest modification time
    most_recent_file = max(files, key=os.path.getmtime)
    return most_recent_file
```

```python
import cv2
import numpy as np
# Example usage:
input_video_path = '/home/linux/DeepLearning/Uday/MP_AVDR/model_training/NumberPlate_Front/Lane_Monitoring/27-05-2024/Lane1-20240527T133917Z-001/Lane1/ch0006_20240527T094840Z_20240527T102322Z_X00000000328000000_1.mp4'
output_video_path = '/home/linux/DeepLearning/Uday/MP_AVDR/model_training/NumberPlate_Front/Lane_Monitoring/27-05-2024/Lane1-20240527T133917Z-001/Lane1/two_vehicle_lane1_video.mp4'


# Define video path and trimming points (in seconds)
video_path = input_video_path
start_time = 6*60
end_time = 7*60  # Exclusive (not including end_time)

# Load video capture object
cap = cv2.VideoCapture(video_path)

# Check if video opened successfully
if not cap.isOpened():
    print("Error opening video file!")
    exit()

# Get video properties (assuming constant frame rate and resolution)
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Calculate frame count for trimming based on timestamps and FPS
num_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
start_frame = int(start_time * fps)
cap.set(cv2.CAP_PROP_POS_FRAMES, start_frame)
end_frame = min(int(end_time * fps), num_frames)  # Handle potential exceeding total frames

# Create video writer for trimmed video (adjust codec and parameters as needed)
fourcc = cv2.VideoWriter_fourcc(*'XVID')  # Adjust codec based on your needs
output_path = output_video_path
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# Read frames and write trimmed video
print('start_frame : ',start_frame)
print('end_frame : ',end_frame)

frame_count = 0+start_frame
while True:
    ret, frame = cap.read()

    if not ret:
        break


    frame_count += 1
    if frame_count%100==0:   
        print("frame_count :  ",frame_count)
    out.write(frame)
    if  frame_count>=end_frame:
        out.write(frame)
        break

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print("Video trimmed and saved successfully!")
```
